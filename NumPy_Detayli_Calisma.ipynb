### 01.1 NumPy dizisi ve temel özellikleri

#### NumPy dizisi nedir?

`ndarray`, değerleri eksenler boyunca düzenleyen NumPy yapısıdır. Bu derste sayısal diziler kullanacağız.
Bir tabloyu `(gözlem, özellik)` düzeninde kurabiliriz; satır ve sütunların anlamını biz belirleriz.

#### Python listesinden farkı

`[10,20] * 2` listeyi tekrar eder. NumPy dizisinde `* 2` her sayıyı ikiyle çarpar.
Böylece sayısal formülleri doğrudan bir veri grubuna uygulayabiliriz.

#### shape, ndim, size ve dtype

| Özellik | Ne anlatır? | `(3,2)` tablo örneği |
|---|---|---|
| `shape` | Her eksenin uzunluğu | 3 satır, 2 sütun |
| `ndim` | Eksen sayısı | 2 |
| `size` | Toplam değer sayısı | 6 |
| `dtype` | Depolama tipi | Örneğin int64 |

Bu özellikler metot değildir; `x.shape()` yerine `x.shape` yazılır.
Bir sayıyı diziye eklemek bütün elemanlara uygulanır. Sonucu atamazsan asıl dizi değişmez.

#### Çözümlü örnek — 01A

**Soru:** Üç öğrencinin iki sınav notunu temsil et. Yapısını incele ve bütün notlara 5 puan ekle.

**Çözüm planı:**

1. Veriyi satırları öğrenci, sütunları sınav olacak şekilde yerleştir.
2. shape, ndim ve size değerlerini oku.
3. Skaler eklemeyi uygulayıp asıl veriyi koruduğunu gözle.

````python
liste_01 = [10, 20]
notlar_01 = np.array([[60, 80], [75, 85], [90, 70]], dtype=np.int64)
print("Liste * 2:", liste_01 * 2)
print("Dizi * 2:", np.array(liste_01) * 2)
print("Notlar:\n", notlar_01)
print("shape:", notlar_01.shape, "ndim:", notlar_01.ndim)
print("size:", notlar_01.size, "dtype:", notlar_01.dtype)
print("5 puan eklenmiş:\n", notlar_01 + 5)
assert notlar_01.shape == (3, 2)
assert notlar_01.size == 6
````

**Çözümün mantığı ve sonucu**

Önce her iç listeyi bir öğrenci olarak yerleştirdik. Üç satır ve iki sütun `(3, 2)` demektir.
Toplam `3 × 2 = 6` değer vardır. Skaler `5`, her elemana uygulanır ve sonuç yine `(3, 2)` olur.
`notlar_01 + 5` yeni bir sonuç üretir; atama yapmadığımız için asıl notları değiştirmedik.

### 01.2 Vektör, satır ve sütun arasındaki fark

#### Tek boyutlu vektör: `(n,)`

`np.array([5,10])` yalnızca bir eksen içerir. `(2,)` gösterimindeki virgül, tek elemanlı tuple yazımıdır.
Bu yapıyı kendiliğinden bir satır veya sütun tablosu saymayacağız.

#### İki boyutlu satır ve sütun

`[[5,10]]` bir satır ve iki sütundur: `(1,2)`. `[[5],[10]]` iki satır ve bir sütundur: `(2,1)`.
Üç temsilde de iki sayı bulunur; eksen düzenleri farklıdır.

#### Bu ayrım neden önemli?

Aynı eleman sayısı, aynı yapı anlamına gelmez. Sonraki bölümlerde satıra veya sütuna göre işlem yaparken
bu düzen sonucu belirleyecek. Önce aşağıdaki küçük örnekte parantezlerin şekle etkisini gözle.

#### Çözümlü örnek — 01B

**Soru:** İki ölçümü vektör, tek satır ve tek sütun olarak göster.

**İzleyeceğin sıra:** Girdiyi ve hedefi belirle → uygun işlemi uygula → sonucu kontrol et.

````python
v_b01 = np.array([5, 10])
satir_b01 = np.array([[5, 10]])
sutun_b01 = np.array([[5], [10]])
for dizi in (v_b01, satir_b01, sutun_b01):
    print("shape:", dizi.shape, "ndim:", dizi.ndim, "size:", dizi.size)
assert v_b01.shape == (2,)
assert satir_b01.shape == (1, 2) and sutun_b01.shape == (2, 1)
````

**Adım adım yorum:** Birinci dizinin bir ekseni, diğerlerinin iki ekseni var. Değer adedi aynı kalsa da satır/sütun düzeni farklıdır.

### 02.1 Dizi oluşturma yöntemleri

#### Hazır değerlerden veya sabitlerden üretme

`np.array` hazır bir listeyi diziye dönüştürür. `zeros` sıfır, `ones` bir,
`full` ise seçtiğin sabit değerle doldurur. Örneğin `np.full((2,3),7)` iki satır ve üç sütundur.

#### arange: adımı belirlemek

`np.arange(2,11,2)` 2'den başlayıp ikişer gider; 11 dahil değildir.
Başlangıç, bitiş sınırı ve adım kullanılır. Dahil olmasını istediğin son değeri dikkate al.

`np.linspace(0,1,5)` iki uç dahil beş nokta üretir. Beş nokta dört aralık oluşturur.

Ondalıklı değer saklayacaksan float kullan. Tamsayı diziye 2.8 atarsan kesir kaybolabilir.
`empty` başlangıç değerlerini hazırlamaz; doldurmadan okunmamalıdır. Bu çalışmada sabit doldurulmuş diziler kullanıyoruz.

#### Çözümlü örnek — 02A

**Soru:** 0–1 arasında beş eşit nokta üret. Tamsayı dizisine 2.8 atamanın etkisini göster ve düzelt.

**Çözüm planı:**

1. Nokta sayısı verilen aralık için linspace seç.
2. Tamsayı diziye ondalık atamanın etkisini gözle.
3. Aynı atamayı float dizide yapıp sonuçları karşılaştır.

````python
noktalar_02 = np.linspace(0, 1, 5)
sayilar_02 = np.arange(2, 11, 2)
tam_02 = np.array([1, 2, 3], dtype=np.int64)
tam_02[0] = 2.8
ondalik_02 = np.array([1, 2, 3], dtype=float)
ondalik_02[0] = 2.8
print("Eşit noktalar:", noktalar_02)
print("Adımlı sayılar:", sayilar_02)
print("Tam sayı ataması:", tam_02)
print("Ondalıklı atama:", ondalik_02)
print("Sabit tablo:\n", np.full((2, 3), 7))
assert tam_02[0] == 2
assert np.isclose(ondalik_02[0], 2.8)
````

**Çözümün mantığı ve sonucu**

Beş nokta dört eşit aralık oluşturur: `0, 0.25, 0.5, 0.75, 1`.
Tamsayı dizisinin tipi atamayla kendiliğinden float'a dönüşmez; 2.8 değeri 2 olarak tutuldu.

### 02.2 Veri tipini dönüştürmek ve kesirleri korumak

#### dtype neden değişmez?

Bir tamsayı dizisine kesirli sayı atamak, diziyi otomatik olarak float'a dönüştürmez.
Yeni değeri mevcut depolama tipine uydurur; bu nedenle beklemediğin bilgi kaybı oluşabilir.

#### astype ile dönüşüm

`x.astype(float)` aynı değerleri float olarak içeren bir dizi hazırlar. Önce bu dönüşümü yap,
sonra kesirli değeri ata. Dönüşümden önce kaybolmuş kesirler geri gelmez.

#### Orijinali korumak

Aşağıdaki örnekte dönüştürülen dizi ayrı değişkende tutulur. İlk diziyi ve yeni diziyi birlikte yazdırarak
hem tipi hem değişikliğin etkisini kontrol et. Dar tamsayı tiplerinde aralık aşımını da düşünmelisin.

#### Çözümlü örnek — 02B

**Soru:** Tamsayı fiyatları değiştirmeden float kopyada bir fiyatı güncelle.

**İzleyeceğin sıra:** Girdiyi ve hedefi belirle → uygun işlemi uygula → sonucu kontrol et.

````python
asil_b02 = np.array([5, 8, 12], dtype=np.int64)
ondalik_b02 = asil_b02.astype(float)
ondalik_b02[0] = 5.75
print("Orijinal:", asil_b02, "Yeni:", ondalik_b02)
assert asil_b02[0] == 5 and np.isclose(ondalik_b02[0], 5.75)
````

**Adım adım yorum:** Önce float dizi oluşturduk, sonra kesirli değeri atadık. Böylece kesir kaybolmadı ve orijinal korunmuş oldu.

### 03.1 reshape ve transpose ile tablo düzenlemek

#### reshape: aynı değerleri yeniden gruplamak

`reshape` toplam eleman sayısını korur. 12 değeri `(3,4)` veya `(2,6)` olarak düzenleyebilirsin;
`(5,3)` için 15 değer gerekir. Önce hedef şeklin boyutlarını çarp.

#### Bilinmeyen boyutu -1 ile hesaplatmak

`reshape(3,-1)` üç satır kurar ve kalan uzunluğu hesaplar. Aynı işlemde yalnızca bir boyut `-1` olabilir.
Eleman sayısı uygun bölünmüyorsa hata alırsın.

#### Transpose: eksenlerin yerini değiştirmek

İki boyutta `.T`, satır ve sütunu takas eder. `(3,4)` tablo `(4,3)` olur.

### 05.1 Boolean maske ve birleşik koşullar

Ve için `&`, veya için `|`, değil için `~` kullanılır. Her karşılaştırmayı paranteze al:
`(yas >= 18) & (puan >= 80)`. Python'un `and` ve `or` operatörleri bu eleman bazındaki iş için uygun değildir.

#### Değerleri ve konumları ayırmak

`x[maske]` değerleri verir; `np.flatnonzero(maske)` tek boyutlu maskede doğru konumların indekslerini verir.
Bir tabloya eleman maskesi uygularsan seçilen değerler tek boyuta inebilir. Satır seçmek için satır başına bir koşul gerekir.

#### Çözümlü örnek — 05A

````python
fiyat_05 = np.array([80, 120, 160, 220, 140])
stok_05 = np.array([3, 0, 5, 2, 4])
maske_05 = (fiyat_05 >= 100) & (fiyat_05 <= 200) & (stok_05 > 0)
secili_05 = fiyat_05[maske_05]
etiket_05 = np.where(stok_05 > 0, "var", "yok")
print("Maske:", maske_05)
print("Seçilen fiyatlar:", secili_05)
print("Stok etiketleri:", etiket_05)
np.testing.assert_array_equal(secili_05, [160, 140])
````

**Çözümün mantığı ve sonucu**

120 uygun fiyatlı olsa da stoku sıfırdır; seçilmez. 160 ve 140 iki koşulu da karşılar.

#### Ondalıklı sayılarda yakınlık

`0.1+0.2` ile `0.3` tam eşit görünmeyebilir; sayılar ikili sistemde yaklaşık saklanır.
`np.isclose` tek tek, `np.allclose` bütün dizi için yakınlığı kontrol eder. Toleransı verinin ölçeğine göre seç.

#### Çözümlü örnek — 07A

````python
fiyat_07 = np.array([100., 200., 300.])
sonuc_07 = fiyat_07 * (1 - 0.10) + 5
print("Yeni fiyatlar:", sonuc_07)
print("Tam eşitlik:", 0.1 + 0.2 == 0.3)
print("Yaklaşık eşitlik:", np.isclose(0.1 + 0.2, 0.3, rtol=1e-10, atol=1e-12))
np.testing.assert_allclose(sonuc_07, [95., 185., 275.])
````

### 08.1 İndirgeme, axis ve keepdims

#### İndirgeme ne yapar?

Toplam ve ortalama gibi işlemler birden fazla değeri özete dönüştürür.
`axis=None` bütün tabloyu birlikte özetler; belirli axis ise o eksen boyunca hesap yapar.

#### Hangi eksen kayboluyor?

`(öğrenci,ders)` tablosunda `axis=0` öğrencileri indirger, ders başına sonuç bırakır.
`axis=1` dersleri indirger, öğrenci başına sonuç bırakır. Eksen numarasını bu mantıkla belirle.

#### keepdims neden var?

`keepdims=True` indirgenen ekseni uzunluğu 1 olarak tutar. `(4,3)` tabloda satır ortalaması
`(4,)` yerine `(4,1)` olur. Sonraki broadcasting bölümünde bu şeklin yararını göreceğiz.

#### Çözümlü örnek — 08A

**Soru:** Üç öğrencinin iki ders notu için ders ve öğrenci ortalamalarını hesapla; sonuç şekillerini göster.

**Çözüm planı:**

1. Her eksenin anlamını yaz.
2. Dersler için öğrenci eksenini, öğrenciler için ders eksenini indir.
3. keepdims ile korunan uzunluğu 1 olan ekseni incele.

````python
notlar_08 = np.array([[50., 70.], [80., 90.], [60., 100.]])
ders_ort_08 = notlar_08.mean(axis=0)
ogrenci_ort_08 = notlar_08.mean(axis=1, keepdims=True)
print("Ders ortalamaları:", ders_ort_08, ders_ort_08.shape)
print("Öğrenci ortalamaları:", ogrenci_ort_08, ogrenci_ort_08.shape)
np.testing.assert_allclose(ders_ort_08, [190/3, 260/3])
np.testing.assert_allclose(ogrenci_ort_08, [[60.], [85.], [80.]])
assert ogrenci_ort_08.shape == (3, 1)
````

**Çözümün mantığı ve sonucu**

axis=0 üç öğrenciyi indirger; iki ders sonucu kalır: `(2,)`. axis=1 iki dersi indirger;
her öğrenciye bir ortalama düşer. keepdims=True sonucu `(3,1)` olarak korur.
Bu şeklin başka bir tabloyla işlemlerde nasıl kullanılacağını bir sonraki bölümde öğreneceğiz.

### 11.1 İstatistiksel özetler ve NaN ile eksik ölçümler

#### Ortalama ve medyan

Ortalama bütün değerlerden etkilenir. Medyan sıralı ortadaki değere dayanır.
Bir uç değer bu iki özeti farklı etkileyebilir; fark görmek uç değeri otomatik silmek için gerekçe değildir.

#### Yayılım ve ddof

Standart sapma yayılımı özetler. `std(ddof=0)` paydada n kullanır; `ddof=1` ile n-1 kullanılır.
Örnek sayısını ve hangi istatistiği hesapladığını dikkate al. Yüzdelikler dağılımın farklı noktalarını gösterir.

#### NaN bulmak ve ortalamadan dışlamak

`np.isnan(x)` eksik konumları bulur; `x==np.nan` kullanma.
`np.nanmean` NaN değerleri dışlayarak hesaplar. Tamamı eksik sütunda öğrenilecek ortalama yoktur.

#### Eksik doldurmanın varsayımı

Sütun ortalamasıyla doldurmak bir yöntemdir; veriye uygunluğu ayrıca değerlendirilmelidir.
ML bağlamında bu ortalama yalnızca eğitim verisinden öğrenilir.

#### Çözümlü örnek — 11A

**Soru:** İki sensörün ölçümlerindeki eksikleri sütun ortalamasıyla doldur; uç değerin etkisini incele.

**Çözüm planı:**

1. Eksik konumları tespit et.
2. Geçerli değerlerden sütun ortalamasını hesapla.
3. Eksikleri doldur ve uç değerin ortalama/medyan etkisini karşılaştır.

````python
olcum_11 = np.array([[10., np.nan], [12., 20.], [np.nan, 22.], [14., 24.]])
ort_11 = np.nanmean(olcum_11, axis=0)
temiz_11 = np.where(np.isnan(olcum_11), ort_11, olcum_11)
uc_11 = np.array([10.,11.,12.,13.,100.])
print("Sütun ortalamaları:", ort_11)
print("Doldurulmuş:\n", temiz_11)
print("Uç değerli ortalama ve medyan:", uc_11.mean(), np.median(uc_11))
print("Std ddof=0 / ddof=1:", np.std(uc_11), np.std(uc_11, ddof=1))
assert not np.isnan(temiz_11).any()
np.testing.assert_allclose(ort_11, [12.,22.])
````

**Çözümün mantığı ve sonucu**

Her sütunda üç geçerli ölçüm var; ortalamalar 12 ve 22. where yalnızca eksik konumları değiştirir.
Uç değerli dizide ortalama 29.2, medyan 12'dir. Farkın nedeni 100'ün ortalamaya güçlü katkısıdır.

### 14.1 Matris çarpımı ve doğrusal tahmin

#### Eleman çarpımı ile matris çarpımı

`*` elemanları çarpar. İki boyutlu dizilerde `@` matris çarpımı yapar.
`(n,p) @ (p,k)` sonucu `(n,k)` olur; iç uzunluklar eşleşmelidir.

#### Özelliklerden tahmine

`X @ w + b` ifadesinde X örnekleri ve özellikleri, w özellik ağırlıklarını, b sabit terimi temsil eder.
`(n,p) @ (p,)` n örnek için `(n,)` tahmin verir. Özelliklerin sırası ağırlıklarla aynı olmalıdır.

#### Sonucu bağımsız kontrol etmek

İlk satırın tahminini elle hesapla. Örnekte ayrıca `solve(A,b)` ile bir denklem çözüp
`A @ çözüm` sonucunu sağ tarafla karşılaştıracağız; bu yöntemi sonraki alt konuda açacağız.

#### Çözümlü örnek — 14A

**Soru:** İki özellikli üç örneğe doğrusal tahmin yap; ayrıca küçük bir denklem sistemini çöz.

**Çözüm planı:**

1. Özellik sayısıyla ağırlık sayısının eşleştiğini kontrol et.
2. Matris çarpımına sabit terimi ekle.
3. Denklem çözümünü yerine koyarak doğrula.

````python
X_14 = np.array([[1.,2.],[2.,1.],[3.,4.]])
w_14 = np.array([2.,3.])
tahmin_14 = X_14 @ w_14 + 1
A_14 = np.array([[2.,1.],[1.,-1.]])
b_14 = np.array([5.,1.])
cozum_14 = np.linalg.solve(A_14, b_14)
print("Tahminler:", tahmin_14)
print("Denklem çözümü:", cozum_14)
print("Geri yerine koyma:", A_14 @ cozum_14)
np.testing.assert_allclose(tahmin_14, [9,8,19])
np.testing.assert_allclose(A_14 @ cozum_14, b_14)
````

**Çözümün mantığı ve sonucu**

İlk tahmin `1×2 + 2×3 + 1 = 9` olur. Üç satır için üç tahmin elde ederiz.
Denklem sisteminin çözümü x=2, y=1'dir. Çözümü A ile çarparak sağ tarafı yeniden ürettik;
sayısal yöntemlerin sonucunu böyle bağımsız bir koşulla kontrol edebiliriz.

### 14.2 Denklem sistemi çözmek ve vektör normu

#### Ax=b biçimi

Her denklemin katsayılarını A'nın bir satırına, sağ tarafını b'ye yaz.
Bütün denklemlerde bilinmeyenlerin sütun sırası aynı kalmalı.

#### solve ile çözüm

`np.linalg.solve(A,b)` kare ve tekil olmayan sistemler için çözüm üretir.
Gereksiz yere matrisin tersini hesaplamak yerine doğrudan bu işlemi kullanabiliriz.
Sonucu yerine koyarak doğrula.

#### Vektörün uzunluğu

`np.linalg.norm(v)` varsayılan tek boyutlu kullanımda Öklid normunu verir.
[3,4] için sonuç sqrt(3²+4²)=5'tir. Norm ile bir denklem çözümü farklı amaçlara hizmet eder.

#### Çözümlü örnek — 14B

**Soru:** x+y=4, x-y=2 sistemini çöz; [6,8] vektörünün uzunluğunu bul.

**İzleyeceğin sıra:** Girdiyi ve hedefi belirle → uygun işlemi uygula → sonucu kontrol et.

````python
A_b14 = np.array([[1., 1.], [1., -1.]])
b_b14 = np.array([4., 2.])
cozum_b14 = np.linalg.solve(A_b14, b_b14)
norm_b14 = np.linalg.norm(np.array([6., 8.]))
print(cozum_b14, norm_b14)
np.testing.assert_allclose(A_b14 @ cozum_b14, b_b14)
np.testing.assert_allclose(cozum_b14, [3., 1.])
assert np.isclose(norm_b14, 10.)
````

**Adım adım yorum:** Katsayı matrisinin sütunları sırasıyla x ve y içindir. Çözümü geri çarparak doğruladık. Norm sqrt(6²+8²)=10 olur.

Şimdi aynı yöntemi aşağıdaki farklı veri üzerinde uygula.

#### Sen çöz — 14B

### 16.1 Eğitimden öğrenmek: fit ve transform

#### Parametre öğrenmek

Ortalama ve standart sapma gibi ön işleme değerleri eğitim verisinden öğrenilir.
Testi bu hesaba katmak eğitim sürecine test bilgisi taşır.

#### Aynı dönüşümü uygulamak

Eksikleri eğitim ortalamasıyla doldurur, ardından `(x-ortalama)/ölçek` uygularız.
Testten ayrı bir ortalama öğrenmek eğitimle farklı bir ölçek oluşturur.

#### Sabit sütunu yönetmek

Standart sapması sıfır sütunda ölçeği 1 seçerek sıfıra bölmeyi önleyeceğiz.
Test ortalamasının sıfır olması gerekmez; eğitim referansına göre konumunu ölçüyoruz.

#### Fonksiyonları okumak

Örnekte öğrenme fonksiyonu parametre sözlüğü döndürür; uygulama fonksiyonu bu sözlüğü kullanır.
Boş veri, sonsuz değer ve tamamı eksik sütunlar için açık kontroller de bulunur.

#### Çözümlü örnek — 16A

**Soru:** Eksik ve sabit sütun içeren eğitim verisinden dönüşüm öğren; test verisini aynı parametrelerle dönüştür.

**Çözüm planı:**

1. Öğrenme fonksiyonunda girdiyi ve eksik sütunları denetle.
2. Eğitimden ortalama ve güvenli ölçek öğren.
3. Aynı parametreleri eğitim ve teste uygulayıp sonuçları kontrol et.

````python
def on_isleme_ogren(X):
    X = np.asarray(X, dtype=float)
    if X.ndim != 2 or X.shape[0] == 0 or X.shape[1] == 0:
        raise ValueError("Boş olmayan iki boyutlu veri gerekli.")
    if np.isinf(X).any():
        raise ValueError("Sonsuz ölçüm kabul edilmiyor.")
    if np.isnan(X).all(axis=0).any():
        raise ValueError("Tamamı eksik sütun için ayrı politika gerekli.")
    ortalama = np.nanmean(X, axis=0)
    dolu = np.where(np.isnan(X), ortalama, X)
    std = dolu.std(axis=0, ddof=0)
    olcek = np.where(std == 0, 1., std)
    return {"ortalama": ortalama, "olcek": olcek}

def on_isleme_uygula(X, parametre):
    X = np.asarray(X, dtype=float)
    if X.ndim != 2 or X.shape[1] != parametre["ortalama"].size:
        raise ValueError("Özellik sayısı eğitimle aynı olmalı.")
    if np.isinf(X).any():
        raise ValueError("Sonsuz ölçüm kabul edilmiyor.")
    dolu = np.where(np.isnan(X), parametre["ortalama"], X)
    return (dolu - parametre["ortalama"]) / parametre["olcek"]

egitim_16 = np.array([[1.,10.,5.],[3.,np.nan,5.],[5.,30.,5.]])
test_16 = np.array([[7.,40.,5.],[np.nan,50.,5.]])
parametre_16 = on_isleme_ogren(egitim_16)
Z_egitim_16 = on_isleme_uygula(egitim_16, parametre_16)
Z_test_16 = on_isleme_uygula(test_16, parametre_16)
print("Öğrenilen:", parametre_16)
print("Eğitim:\n", Z_egitim_16)
print("Test:\n", Z_test_16)
np.testing.assert_allclose(Z_egitim_16.mean(axis=0), 0, atol=1e-12)
np.testing.assert_allclose(Z_egitim_16[:, 2], 0, atol=1e-12)
assert np.isfinite(Z_test_16).all()
````

**Çözümün mantığı ve sonucu**

Ortalamalar `[3,20,5]` yalnızca eğitimden öğrenildi. Sabit üçüncü sütunun ölçeği 1 oldu ve
eğitimde sıfıra dönüştü. Testin ortalamasının sıfır olması gerekmez; eğitim referansına göre
konumunu görmek istiyoruz. Testte yeni bir değer sabit sütundan farklı gelirse sonuç sıfır olmak

### 18.1 NumPy ile regresyon ve gradyan inişi

#### Veri üretimi ve bölme

#### Tasarım matrisi ve parametreler

#### Kayıp ve gradyan

`mean((tahmin-y)**2)` ortalama karesel hatadır. Gradyan bu kaybın parametreler değişirken
hangi yönde arttığını anlatır. Bu örnekte `(2/n)*tasarım.T@(tahmin-y)` kullanılır.

#### Güncelleme ve değerlendirme

`theta -= öğrenme_oranı*gradyan` ile kaybı azaltmaya çalışırız. 0.1 bu küçük ölçeklenmiş
problem için seçilir; her veride uygun değildir. En son testte eğitim hedef ortalamasını kullanan
referansla karşılaştırırız. MAE mutlak hataların ortalaması, RMSE karesel hatanın kareköküdür.

#### Çözümlü örnek — 18A

**Soru:** Yapay veri üret, gradyan inişiyle modeli eğit; test hatasını referans modelle karşılaştır.

**Çözüm planı:**

1. Yapay veriyi üret ve eğitim/test indekslerini ayır.
2. Eğitimden öğrenilen ölçekle tasarım matrisini kur.
3. Gradyan inişiyle eğit; en son testte referans modelle karşılaştır.

````python
rng_18 = np.random.default_rng(2026)
X_18 = rng_18.normal(size=(160,2))
y_18 = 3 + X_18 @ np.array([4.,-2.]) + rng_18.normal(0,0.15,size=160)
indeks_18 = rng_18.permutation(len(y_18))
egitim_idx_18, test_idx_18 = indeks_18[:128], indeks_18[128:]
X_egitim_18, X_test_18 = X_18[egitim_idx_18], X_18[test_idx_18]
y_egitim_18, y_test_18 = y_18[egitim_idx_18], y_18[test_idx_18]
p_18 = on_isleme_ogren(X_egitim_18)
Z_egitim_18 = on_isleme_uygula(X_egitim_18,p_18)
Z_test_18 = on_isleme_uygula(X_test_18,p_18)
tasarim_18 = np.column_stack([np.ones(len(Z_egitim_18)), Z_egitim_18])
test_tasarim_18 = np.column_stack([np.ones(len(Z_test_18)), Z_test_18])
theta_18 = np.zeros(tasarim_18.shape[1])
kayiplar_18 = []
for adim_18 in range(800):
    hata_18 = tasarim_18 @ theta_18 - y_egitim_18
    kayiplar_18.append(np.mean(hata_18**2))
    gradyan_18 = (2 / len(y_egitim_18)) * tasarim_18.T @ hata_18
    theta_18 -= 0.1 * gradyan_18
tahmin_18 = test_tasarim_18 @ theta_18
mse_18 = np.mean((tahmin_18 - y_test_18)**2)
referans_18 = np.mean((y_egitim_18.mean() - y_test_18)**2)
orijinal_w_18 = theta_18[1:] / p_18['olcek']
orijinal_b_18 = theta_18[0] - p_18['ortalama'] @ orijinal_w_18
print("Başlangıç / son kayıtlı eğitim kaybı:", kayiplar_18[0], kayiplar_18[-1])
print("Test MSE / referans MSE:", mse_18, referans_18)
print("Orijinal ölçekte w ve b:", orijinal_w_18, orijinal_b_18)
assert kayiplar_18[-1] < kayiplar_18[0]
assert mse_18 < referans_18
````

**Çözümün mantığı ve sonucu**

Tasarım matrisi `(128,3)`, parametreler `(3,)` ve eğitim tahmini `(128,)` şeklindedir.
Üç parametre bir sabit ve iki özellik katsayısıdır. Ölçeklenmiş uzayda öğrendiğimiz parametreleri

### 18.2 En küçük kareler ile bağımsız regresyon çözümü

#### Aynı problemi farklı yöntemle çözmek

Gradyan inişi adım adım ilerler. `np.linalg.lstsq` doğrusal sistemde toplam karesel hatayı
minimum yapan parametreleri sayısal olarak bulur.

#### Sabit ve eğim sütunları

Tek özellikli problemde birler sütunu ve x sütununu yan yana koyarız.
İlk parametre sabit, ikinci parametre eğim olur. `rcond=None` ile örnekteki standart çağrıyı yapacağız.

#### Yeni veri için tahmin

Yeni girdide aynı sütun sırasını kurup öğrenilen parametrelerle çarp.
Gürültüsüz veride bilinen doğruyu yaklaşık olarak geri kazanabiliriz; sayısal karşılaştırmayı toleransla yap.

#### Çözümlü örnek — 18B

**Soru:** y=1+2*x ilişkisini gradyan inişi yerine lstsq ile öğren.

**İzleyeceğin sıra:** Girdiyi ve hedefi belirle → uygun işlemi uygula → sonucu kontrol et.

````python
x_b18 = np.array([-1., 0., 1., 2.])
y_b18 = 1 + 2 * x_b18
tasarim_b18 = np.column_stack([np.ones(len(x_b18)), x_b18])
theta_b18, _, _, _ = np.linalg.lstsq(tasarim_b18, y_b18, rcond=None)
yeni_b18 = np.array([3., 5.])
yeni_tasarim_b18 = np.column_stack([np.ones(len(yeni_b18)), yeni_b18])
tahmin_b18 = yeni_tasarim_b18 @ theta_b18
print("Sabit ve eğim:", theta_b18, "Tahmin:", tahmin_b18)
np.testing.assert_allclose(theta_b18, [1., 2.], atol=1e-12)
np.testing.assert_allclose(tahmin_b18, [7., 11.], atol=1e-12)
````

**Adım adım yorum:** Birler sütunu sabit terimi, x sütunu eğimi temsil eder. lstsq toplam karesel hatayı en aza indiren parametreleri bulur. Bu gürültüsüz veride ilişki tam olarak geri kazanılır.

Şimdi aynı yöntemi aşağıdaki farklı veri üzerinde uygula.
